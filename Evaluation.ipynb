Import Libraries

```python
import pandas as pd
import numpy as np
```

Load Model Outputs

```python
department_summary = pd.read_csv("department_summary.csv")

category_summary = pd.read_csv("category_summary.csv")

monthly_summary = pd.read_csv("monthly_summary.csv")
```

Evaluate Overall Company Performance
Total Budget

```python
total_budget = department_summary['Budget Amount'].sum()

print("Total Budget:", total_budget)
```

Total Actual Spending

```python
total_actual = department_summary['Actual Amount'].sum()

print("Total Actual Spending:", total_actual)
```

Total Variance

```python
total_variance = department_summary['Variance'].sum()

print("Total Variance:", total_variance)
```

Business Interpretation
Positive Variance = Overspending

Negative Variance = Savings

Best Performing Department

Departments closest to their budgets.

```python
best_department = department_summary.sort_values(
    by='Variance',
    ascending=True
).head(1)

best_department
```

Worst Performing Department

```python
worst_department = department_summary.sort_values(
    by='Variance',
    ascending=False
).head(1)

worst_department
```

Department Ranking Analysis

```python
department_summary.sort_values(
    by='Department Rank'
)
```

Highest Overspending Category

```python
worst_category = category_summary.sort_values(
    by='Variance',
    ascending=False
).head(1)

worst_category
```

Best Expense Category

```python
best_category = category_summary.sort_values(
    by='Variance',
    ascending=True
).head(1)

best_category
```

Monthly Performance Analysis
Best Month

```python
best_month = monthly_summary.sort_values(
    by='Efficiency Score',
    ascending=False
).head(1)

best_month
```

Worst Month

```python
worst_month = monthly_summary.sort_values(
    by='Efficiency Score',
    ascending=True
).head(1)

worst_month
```

Budget Utilization Analysis
Department Utilization

```python
department_summary[
    ['Department',
     'Budget Utilization %']
].sort_values(
    by='Budget Utilization %',
    ascending=False
)
```

Risk Analysis
High-Risk Departments

```python
high_risk_departments = department_summary[
    department_summary['Budget Utilization %'] > 110
]

high_risk_departments
```

Savings Opportunity Analysis

```python
savings_opportunities = category_summary[
    category_summary['Variance'] < 0
]

savings_opportunities
```

Generate Key Findings

Create a summary section:

```python
print("Key Findings")
print("="*50)

print("Best Department:")
print(best_department['Department'].values[0])

print("Worst Department:")
print(worst_department['Department'].values[0])

print("Highest Overspending Category:")
print(worst_category['Category'].values[0])

print("Best Month:")
print(best_month['Month'].values[0])

print("Worst Month:")
print(worst_month['Month'].values[0])
```

Visualization for Final Report
Department Ranking

```python
import matplotlib.pyplot as plt

department_summary.sort_values(
    by='Variance'
).plot(
    x='Department',
    y='Variance',
    kind='bar',
    figsize=(10,5)
)

plt.title("Department Variance Ranking")
plt.show()
```

Category Ranking

```python
category_summary.sort_values(
    by='Variance'
).plot(
    x='Category',
    y='Variance',
    kind='bar',
    figsize=(10,5)
)

plt.title("Category Variance Ranking")
plt.show()
```
